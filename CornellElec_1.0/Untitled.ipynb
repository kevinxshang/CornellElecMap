```python
import json

with open('cugir-008163-geojson_copy.json', 'r') as file:
    json_data = json.load(file)
    
def filter_by_municipality(json_data, municipality):
    filtered_data = {
        "type": "FeatureCollection",
        "features": []
    }
    for feature in json_data['features']:
        if feature['properties']['municipali'] == municipality:
            filtered_data['features'].append(feature)
    return filtered_data

# Filter data for "CORNELL UNIVERSITY"
filtered_data = filter_by_municipality(json_data, "CORNELL UNIVERSITY")

# Save the filtered data to a new JSON file
with open('filtered_cornell_data.json', 'w') as outfile:
    json.dump(filtered_data, outfile, indent=4)

filtered_data  # Displaying the filtered data for reference
```

```python
with open('filtered_cornell_data.json', 'r') as file:
    json_data = json.load(file)

# Function to extract unique building names
def extract_unique_building_names(json_data):
    building_names = set()
    for feature in json_data:
        building_name = feature['properties']['buildingna']
        if building_name:
            building_names.add(building_name)
    return building_names

# Extract unique building names from the loaded JSON data
unique_building_names = extract_unique_building_names(json_data['features'])  # assuming 'features' is the key containing the list of buildings
print(unique_building_names)
```
